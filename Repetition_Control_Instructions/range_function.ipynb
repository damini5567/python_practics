### range()

#### Syntax
##### range(start,stop,step)

```python
for i in range(10):
    print(i)
```

```python
for i in range(1,6):
    print(i)
```

```python
for i in range(0,10,1):
    print(i)
```

```python
for i in range(0,10,2):
    print(i)
```

```python
for i in [0,1,2,3,4,5]:
    print(i)
```
