### REPETITION CONTROL INSTRUCTIONS

### 2) for loop

```python
contain="PYTHON"
for char in contain:
    print(char)
print("python is a programming language.")
```

#### i) List

```python
numbers = [1, 2, 3, 4, 5]
for i in numbers:
    print(i)
```

#### ii) Tuple

```python
numbers = (1, 2, 3, 4, 5)
for i in numbers:
    print(i)
```

#### iii) Set

```python
print("FRUITS")
fruits = {"Apple","Mango","Grapes","Orange"}
for fruit in fruits:
    print(fruit)
```

#### iv) Dictionary
##### By default, a for loop iterates over the keys of a dictionary.

```python
student = {
    "name": "Avantika",
    "city": "Pune"
}
for key in student:
    print(key)
```
