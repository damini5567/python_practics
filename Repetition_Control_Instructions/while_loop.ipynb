### REPETITION CONTROL INSTRUCTIONS

### 1) while loop

```python
i = 1
while (i<=10):
    print(i)
    i=i+1
```

```python
count=0
while (count<=5):
    print(count)
    count=count+1
print("Program Executed")
```

```python
n = 1
count = 0
while n <= 20:
    if n % 2 == 0:
        count += 1
    n += 1
print(count)
```
