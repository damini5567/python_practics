## Set Built-in functions

### 1.len( )

```python
set1={10,20,30,40,50}
print(len(set1))
```

### 2. max()

```python
set1={10,20,30,40,50}
print(max(set1))
```

### 3.min()

```python
set1={10,20,30,40,50}
print(min(set1))
```

### 4.sum()

```python
set1={10,20,30}
print(sum(set1))
```

### 5.any()

```python
set1={0,0,0,0,1}
print(any(set1))
```

### 6.all()

```python
set1={10,20,30,40,50,60}
print(all(set1))
```

```python
set2={10,20,30,40,50,0}
print(all(set2))
```
