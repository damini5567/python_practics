## Set Methods

### 1. add()

```python
set1={10,20,30}
set1.add(40)
print(set1)
```

### 2. update()

```python
set1={10,20,30}
set2={40,50}
set1.update(set2)
print(set1)
```

### 3.remove()

```python
set1={10,20,30,40,50}
set1.remove(30)
print(set1)
```

```python
set1.remove(100)
```

### 4.discard()

```python
set1={10,20,30,40,50}
set1.discard(30)
print(set1)
```

```python
set1.discard(100)
```

### 5. clear()

```python
set1={10,20,30,40,50}
set1.clear()
print(set1)
```

### 6. copy( )

```python
set1={10,20,30,40,50}
set2=set1.copy()

print(id(set1))
print(id(set2))
```
