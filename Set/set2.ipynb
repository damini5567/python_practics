## Basic set operations

### 1.mutability

```python
set1={10,20,30}
set1.add(40)
print(set1)
```

### 2. conversion

#### 1. converting string to set

```python
str1="hello"
set1=set(str1)
print(set1)
```

#### 2. converting list to set

```python
list1=[10,20,30]
set1=set(list1)
print(set1)
```

#### 3. converting tuple to set

```python
tuple1=(10,20,30)
set1=set(tuple1)
print(set1)
```

#### 4. converting dictionary to set

```python
dict1={"a":1,"b":2}
set1=set(dict1)
print(set1)
```

#### 5.converting range( ) to set

```python
range1=range(1,11,1)
set1=set(range1)
print(set1)
```

### 3. Aliasing

```python
set1={10,20,30}
set2=set1
print(id(set1))
print(id(set2))
```

### 4.cloning

```python
set1={10,20,30}
set2=set1.copy()
print(id(set1))
print(id(set2))
```

### 5.searching - in operator

```python
set1={10,20,30}
print(10 in set1)
print(100 in set1)
```

### 6. identity - is operator

```python
set1={10,20,30}
set2=set1
print(set1 is set2)
```

```python
set3={10,20,30}
set4=set3.copy()
print(set3 is set4)
```

### 7. Emptiness

```python
set1=set()
print(not set1)
```

```python
set2={10,20,30}
print(not set2)
```
