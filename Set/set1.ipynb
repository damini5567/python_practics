## Sets

### homogeneous elements in the sets

```python
set1={10,20,30,40,50}
print(set1)
```

```python
print(type(set1))
```

### hetrogeneous elements in the sets

```python
set2={10,10.1,"Avantika",2+3j,True}
print(set2)
```

```python
print(type(set2))
```

### Duplicate elements are not allowed

```python
set3={10,20,30,40,50,10}
print(set3)
```

### Creation of empty set

```python
set4=set()
print(set4)
```

```python
print(type(set4))
```

### looping in sets

#### 1. for loop

```python
set1={10,20,30,40,50}
for i in set1:
    print(i)
```

#### 2. enumerate ()

```python
set1={10,20,30,40,50}
for index,value in enumerate(set1):
    print(index,value)
```
