## STRINGS

##### 1) Single quotes - ' '


```python
greetings='jay ganesh'
print(greetings)
print(type(greetings))
```

##### 2) Double quotes - " "

```python
greetings1="jay ganesh"
print(greetings1)
print(type(greetings1))
```

##### 3) Tripple quotes - '''  '''    or    """  """

###### i) single line string

```python
greetings3='''jay ganesh'''
print(greetings3)
print(type(greetings3))
```

```python
greetings3="""jay ganesh"""
print(greetings3)
print(type(greetings3))
```

###### ii) Multiline String

```python
greetings4='''Hello 
welcome to AI Course
'''
print(greetings4)
print(type(greetings))
```

```python
greetings4="""Hello 
welcome to AI Course
"""
print(greetings4)
```
