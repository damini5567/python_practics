## STRING METHODS

### 1) CONTENT TEST FUNCTIONS

#### a) isalpha()

```python
str1="Artifical intelligence"
print(str1.isalpha())
```

```python
str1="Artificalintelligence"
print(str1.isalpha())
```

```python
str1="@Artificalintelligence"
print(str1.isalpha())
```

```python
str1="@Artifical_intelligence"
print(str1.isalpha())
```

```python
str1="Artifical_intelligence1"
print(str1.isalpha())
```

### b) isdigit()


```python
str1="12345"
print(str1.isdigit())
```

```python
str1="123 456"
print(str1.isdigit())
```

```python
str1="20.5"
print(str1.isdigit())
```

```python
str2="hello123"
print(str2.isdigit())
```

### c) isalnum()

```python
str1="hello"
print(str1.isalnum())
```

```python
str1="12345"
print(str1.isalnum())
```

```python
str1="hello123"
print(str1.isalnum())
```

```python
str1="hello@"
print(str1.isalnum())
```

```python
str1="hello 123"
print(str1.isalnum())
```

### d) islower()


```python
str1="hello"
print(str1.islower())
```

### e) isupper()

### f) startswith()


```python
str1="Hello welcome"
print(str1.startswith("Hello"))
```

### g) endswith()
