## TRIMS WHITESPACES

#### 1) lstrip()

```python
str1="    Hello,Welcome to python programming"
print(str1)
str2=str1.lstrip()
print(str2)
```

#### 2) rstrip()

```python
str1="Hello,Welcome to python programming    "
print(str1)
str2=str1.rstrip()
print(str2)
```

#### 3) strip()

```python
str1="    Hello,Welcome to python programming    "
print(str1)
str2=str1.strip()
print(str2)
```
