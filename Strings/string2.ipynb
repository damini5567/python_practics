### String Slicing

```python
str1="jai ganesh"
print(str1[0:3:1])
print(str1[4:10:2])
print(str1[4:10:1])
print(str1[0:-6:1])
print(str1[::])
```

#### Reverse The String

```python
str1="string"
print(str1[::-1])
```

#### String Properties

##### 1) Immutability

```python
str1="jai ganesh"
str1[0]="k"
```

##### 2) Concatenation

```python
str1="jai"
str2="ganesh"
str3=" "
str4=str1+str3+str2
print(str4)
```

#### 3) Repetation

```python
str1="Kaveri "
str2=str1*5
print(str2)
```

#### 4) Indexing

```python
str="JAI GANESH"
print(str[1])
```

#### 5) Slicing

```python
str="JAI GANESH"
print(str[1:4:1])
```

#### 6) Membership Testing

```python
str1="python is a programming language"
print("python" in str1)
print("Python" in str1)
```
