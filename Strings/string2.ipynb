### String Slicing

```python
str1="jai ganesh"
print(str1[0:3:1])
print(str1[4:10:2])
print(str1[4:10:1])
print(str1[0:-6:1])
print(str1[::])
```

#### Reverse The String

```python
str1="string"
print(str1[::-1])
```
