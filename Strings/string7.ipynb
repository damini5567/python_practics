### STRING MANIPULATION METHODS

#### 1) split()

```python
str1="Hello, welcome to python"
words=str1.split()
print(words) 
```

```python
str1="Hello, welcome to python"
words=str1.split(" ")
print(words) 
```

#### 2) partition 

```python
str1="Hello, welcome to python course"
before,sep,after=str1.partition("to")
print(before)
print(sep)
print(after)
```
