### BUILT IN FUNCTIONS 

#### 1) str

```python
a=20 
b=20.1
c=20+1j
```

```python
new_a=str(a)
print(new_a)
print(type(new_a))
```

```python
new_b=str(b)
print(new_b)
print(type(new_b))
```

```python
new_c=str(c)
print(new_c)
print(type(new_c))
```

#### 2) int()

```python
d="10"
print(d)
print(type(d))
new_d=int(d)
print(new_d)
print(type(new_d))
```

#### 3) float

```python
e="10.1"
print(e)
print(type(e))
new_e=float(e)
print(new_e)
print(type(new_e))
```

#### 4) complex()

```python
num="20+1j"
print(num)
print(type(num))
new_num=complex(num)
print(new_num)
print(type(new_num))
```

## join() method

```python
words=["My","name","is","Avantika"]
sentence=" ".join(words)
print(sentence)
```

```python
words=["My","name","is","Avantika"]
sentence="-".join(words)
print(sentence)
```

```python
words=["My","name","is","Avantika"]
sentence="|".join(words)
print(sentence)
```

```python
words=["My","name","is","Avantika"]
sentence="_".join(words)
print(sentence)
```
