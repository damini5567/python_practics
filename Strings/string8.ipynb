### STRING CONVERSION METHODS

#### 1) lower()

```python
str1="HELLO WELCOME"
new_str=str1.lower()
print(new_str)
```

#### 2) upper()

```python
str1="Hello welcome"
new_str=str1.upper()
print(new_str)
```

#### 3) capitalize()

```python
str1="hello WELCOME"
new_str=str1.capitalize()
print(new_str)
```

#### 4) title()

```python
str1="hello welcome"
new_str=str1.title()
print(new_str)
```

#### 5) swapcase()

```python
str1="hELLO wELCOME"
new_str=str1.swapcase()
print(new_str)
```

#### 6) casefold()

```python
name1="HELLO"
name2="hello"
print(name1.casefold()==name2.casefold())
```
