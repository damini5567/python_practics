## Built in functions on list

### 1. len ( )

```python
list1=[10,20,30,40,50]
print(list1)
print(type(list1))

print("Length of List is :",len(list1))
```

### 2. max ( )

```python
list1=[10,20,30,40,50]
print(max(list1))
```

### 3. min ( )

```python
list1=[10,20,30,40,50]
print(min(list1))
```

### 4. sum( )

```python
list1=[10,20,30,40,50]
print(sum(list1))
```

### 5.any( )

```python
list1=[0,0,0,0,1]
print(any(list1))

list2=[0,0,0,0,0]
print(any(list2))
```

### 6.all ( )

```python
list1=[10,20,30,40,50]
print(all(list1))

list2=[10,20,30,40,0]
print(all(list2))
```

### 7. sorted( )

```python
list1=[50,40,10,20,30]
print(sorted(list1))
```

### 8.reversed( )

```python
list1=[10,20,30,40,50]
result=reversed(list1) 
print(result)
```
