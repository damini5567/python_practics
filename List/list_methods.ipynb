## List Methods

### Adding Elements Methods

#### 1. append( )

```python
list1=[10,20,30,40,50]
list1.append(60)
print(list1)

list1.append("Avantika")
print(list1)

list1.append(100.50)
print(list1)
```

#### 2. insert ( )

```python
list1=[10,20,30,40,50]
list1.insert(2,25)
print(list1)

list1.insert(0,5)
print(list1)

list1.insert(6,45)
print(list1)

list1.insert(8,55)
print(list1)

list1.append(60)
print(list1)
```

#### 3.extend( )

```python
list1=[10,20,30]
list2=[40,50,60]

list1.extend(list2)

print(list1)
print(list2)
```

### Removing Elements Methods

#### 1. remove ( )

```python
list1=[10,20,30,40,50]
list1.remove(30)
print(list1)
```

```python
list1.append(60)
print(list1)
```

```python
list1.remove(60)
print(list1)
```

### 2. pop ( )

##### first way - pop( )

```python
list1=[10,20,30,40,50]
list1.pop()
print(list1)
```

#### second way - pop (index no)

```python
list1=[10,20,30,40,50]
list1.pop(2)
print(list1)
```

### 3. clear ( )

```python
list1=[10,20,30,40,50]
list1.clear()
print(list1)
```

## 3) searching and counting methods

### 1. index()

```python
list1=[10,20,30,40,50]

print(list1.index(30))

print(list1.index(10))

print(list1.index(50))
```

### 2. count ( )

```python
list1=[10,20,30,10,50,20,10]

print(list1.count(10))

print(list1.count(20))

print(list1.count(30))

print(list1.count(100))
```

## 3) sorting and reversing

### 1. sort( )

```python
list1=[50,40,30,5,45,35,10]
list1.sort()
print(list1)
```

### 2. reverse( )

```python
list1=[10,20,30,40,50]

list2=list1.copy()

print(list1)

print(id(list1))

print(id(list2))
```

### 3. copy( )

```python
list1=[10,20,30,40,50]
list2=list1.copy()

print(list1)

print(id(list1))

print(id(list2))
```
