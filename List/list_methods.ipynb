## List Methods

### Adding Elements Methods

#### 1. append( )

```python
list1=[10,20,30,40,50]
list1.append(60)
print(list1)

list1.append("Avantika")
print(list1)

list1.append(100.50)
print(list1)
```

#### 2. insert ( )

```python
list1=[10,20,30,40,50]
list1.insert(2,25)
print(list1)

list1.insert(0,5)
print(list1)

list1.insert(6,45)
print(list1)

list1.insert(8,55)
print(list1)

list1.append(60)
print(list1)
```

#### 3.extend( )

```python
list1=[10,20,30]
list2=[40,50,60]

list1.extend(list2)

print(list1)
print(list2)
```
