### 5. Aliasing - (Shallow Copy)

##### Aliasing occurs when multiple variables refer to the same list object in memory
##### (means multiple variable store same memory location)
##### when one variable assign to another variable they store in same memory location

```python
list10=[1,2,3,4,5]
print("Before Aliasing list10 :",list10)
print(id(list10))
print("\n")

list9=[6,7,8,9,10]
print("Before Aliasing list9 :",list9)
print(id(list9))
print("\n")

list10=list9       # Aliasing
print("After Aliasing list10=list9 :",list10)
print(id(list10))
print(id(list9))
```

```python
list1=[10,20,30]
print(id(list1))

list2=list1
print(id(list2))

list1[0]=11
print(list1)
print(list2)
```

```python
list3=list1
list4=list1
list5=list1
print(list1)
print(list2)
print(list3)
print(list4)
print(list5)


list1[2]=31
print(list1)
print(list2)
print(list3)
print(list4)
print(list5)
```

```python
list5[1]=21
print(list1)
print(list2)
print(list3)
print(list4)
print(list5)

print(id(list1))
print(id(list2))
print(id(list3))
print(id(list4))
print(id(list5))
```

#### Disadvantage : Changes to one list it will affect on other list also

### 5. cloning / Deep copy
##### It creates separate copy of a list on different memory location
##### Advantage : Change to one list doesn't affect to another

```python
list11=[10,20,30]
print(id(list11))
print(list11)
print("\n")

list12=[40,50,60]
print(id(list12))
print(list12)
print("\n")

list12=list11.copy()
print(id(list12))
print(list12)
```
