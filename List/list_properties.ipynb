## Basic List Operations

### 1) Mutability

##### We can modify string means we can change or remove items, still the memory address is same

```python
List1=[10,20,30,40,50]
print(List1)
print(type(List1))
List1[0]=0
print(List1)
```

### string is immutable

```python
str1="Hello"
print(str1)
print(type(str1))
```

```python
str1[0]="h"
```

### 2.Concatenation and 3.Merging-  + operator

##### We can join two list together to create new combine list.
##### + operator used for concatenation.

```python
list2=[10,20,30]
list3=[40,50,60]
list4=list2+list3
print(list4)
```

```python
list2=[10,20,30]
list3=[40,50,60]

print("Combine list is :",list2+list3)
print("Combine list is :",list2+list3+[70,80,90,100])
```
