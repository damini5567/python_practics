### while loop

```python
list1=[10,20,30,40,50]
index=0
while(index<len(list1)):
    print(list1[index])
    index=index+1
```

### for loop

```python
list1=[10,20,30,40,50]
for var in list1:
    print(var)
```

### enumerate() function

```python
list1=[10,20,30,40,50]
for index,var in enumerate(list1):
    print(index,var)
```
