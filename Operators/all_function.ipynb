### 1) all function

```python
a=1;
b=2;
c=3;
print(all((a<5,b<5,c<5)))
```

```python
age=23
nationality="indian"
voting_card=False
if all ((age>=18,nationality=="indian",voting_card)):
    print("you are eligible for voting")
else:
    print("you are not eligible for voting")
```

### 2) any function

```python
age=23
nationality="American"
voting_card=False
if any((age>=18,nationality=="indian",voting_card)):
    print("you are eligible for voting")
else:
    print("you are not eligible for voting")
```
