### LOGICAL OPERATOR

##### 1) AND OPERATOR

```python
age=23
nationality="Indian"
if (age>=18) and (nationality=="Indian"):
    print("you are eligible for voting")
else:
    print("you are not eligible for voting")
```

##### 2) OR OPERATOR

```python
is_weekend=True
is_holiday=False
if is_weekend or is_holiday:
    print("You can relax")
else:
    print("You need to work")
```
