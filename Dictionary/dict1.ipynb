## Dictionary

### Creation of dictionary

```python
student={
    "name":"Avantika",
    "roll_no":100,
    "address":"pune"
}
```

```python
print(student)
```

```python
print(type(student))
```

### Different keys may have same value

```python
dict2={"name":"Avantika","roll_no":100,"account_bal":100}
print(dict2)
```

### Keys should be unique

```python
dict1={
    "name":"ganesh",
    "roll_no":100,
    "name":"mahesh",
    "roll_no":101
}
```

```python
print(dict1)
```

### Accessing dictionary elements

```python
dict1={
    "name":"Avantika",
    "roll_no":100
}
print(dict1)
```

```python
print(dict1["name"])
```

```python
print(dict1["roll_no"])
```

### Looping in dictionaries

```python
dict1={"name":"Avantika",
        "roll_no":100
    }
```

#### for loop- iterating keys and values only - items()

```python
for k,v in dict1.items():
    print(k,v)
```

#### for loop- iterating keys only - keys()

```python
dict2={"name":"Avantika",
        "roll_no":100
    }
```

```python
for k in dict2.keys():
    print(k)
```

#### for loop- iterating values only - values()

```python
dict3={"name":"Avantika",
        "roll_no":100
    }
```

```python
for v in dict3.values():
    print(v)
```

### enumerate()

```python
dict4={"name":"Avantika",
        "roll_no":100
    }
```

```python
for i,(k,v) in enumerate(dict4.items(),start=10):
    print(i,k,v)
```
