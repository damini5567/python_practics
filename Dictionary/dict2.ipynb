## Basic dictionary operations

### 1. Mutability

```python
dict1={
    "name":"Avantika",
    "roll_no":100
}
```

```python
print(dict1)
```

```python
print(type(dict1))
```

```python
dict1["age"]=25
print(dict1)
```

### 2. conversion

#### 1. convert a tuple(tupe of tuples) into dictionary

```python
tuple1=(("name","Avantika"),("age",25))
print(tuple1)
print(type(tuple1))
```

```python
print(tuple1[0])
print(tuple1[0][0])
print(tuple1[0][1])
```

```python
tuple1[1]
```

```python
print(tuple1)
```

```python
dict1=dict(tuple1)
print(dict1)
```

```python
print(type(dict1))
```
