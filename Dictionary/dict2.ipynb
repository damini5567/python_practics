## Basic dictionary operations

### 1. Mutability

```python
dict1={
    "name":"Avantika",
    "roll_no":100
}
```

```python
print(dict1)
```

```python
print(type(dict1))
```

```python
dict1["age"]=25
print(dict1)
```

### 2. conversion

#### 1. convert a tuple(tupe of tuples) into dictionary

```python
tuple1=(("name","Avantika"),("age",25))
print(tuple1)
print(type(tuple1))
```

```python
print(tuple1[0])
print(tuple1[0][0])
print(tuple1[0][1])
```

```python
tuple1[1]
```

```python
print(tuple1)
```

```python
dict1=dict(tuple1)
print(dict1)
```

```python
print(type(dict1))
```

#### 2. convert a list(list of tuples) into dictionary

```python
list1=[("name","Avantika"),("age",23)]
dict1=dict(list1)
print(dict1)
```

#### 3. convert a set of elements as key and default values as dictionary

```python
set1={"num1","num2","num3"}
dict1={key:10 for key in set1}
print(dict1)
```

#### 4. converting range as keys and values as default to dictionary

```python
range1=range(0,6,1)
dict1={key:10 for key in range1}
print(dict1)
```

#### 3. Aliasing

```python
dict1={"name":"Avantika",
       "age":25}
dict2=dict1
```

```python
print(id(dict1))
print(id(dict2))
```

#### 4. Cloning

```python
dict1={"name":"Avantika",
       "age":25}
dict2=dict1.copy()
```

```python
print(id(dict1))
print(id(dict2))
```

#### 5.searching - in opeartor

```python
dict1={"name":"Avantika",
       "age":25,
       "roll_no":100}
```

```python
print("age" in dict1)
```

```python
print("address" in dict1)
```

#### 6. Identity - is operator

```python
dict1={"name":"Avantika",
       "age":25}
```

```python
dict2=dict1
```

```python
print(id(dict1))
print(id(dict2))
```

```python
print(dict1 is dict2)
```

```python
dict3=dict1.copy()
```

```python
print(id(dict1))
print(id(dict3))
```

```python
print(dict3 is dict1)
```

#### 7.Emptiness - not operator

```python
dict1={}
print(not dict1)
```

```python
dict1={"name":"Avantika"}
print(not dict1)
```
