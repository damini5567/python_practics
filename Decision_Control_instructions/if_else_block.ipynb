### 2) if_else block

```python
a = 25
b = 15
if (a > b):
    print("A is greater")
else:
    print("B is greater")
```

```python
num = 10
if (num >= 0):
    print("Positive")
else:
    print("Negative")
```

```python
num = 150
if (num >= 100):
    print("Number is greater than 100")
else:
    print("Number is less than or equal to 100")
```
