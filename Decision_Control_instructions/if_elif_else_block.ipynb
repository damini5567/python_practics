### 3) if-elif-else block

```python
age=25
if (age<13):
    print("You are child,not eligible for voting")
elif (age<18):
    print("You are teenager,not eligible for voting")    
else:
    print("You are eligible for voting")    
```

```python
number = 10

if number > 0:
    print("Positive number")
elif number == 0:
    print("Zero")
else:
    print("Negative number")
```
