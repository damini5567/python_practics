### 3) if-elif-else block

```python
num = 10
if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")
```

```python
num = 100
if num > 100:
    print("Greater than 100")
elif num == 100:
    print("Equal to 100")
else:
    print("Less than 100")
```

```python
language = "C++"
if language == "Python":
    print("Python selected")
elif language == "Java":
    print("Java selected")
elif language == "C++":
    print("C++ selected")
else:
    print("Unknown language")
```

```python
marks = 85
if marks >= 90:
    print("Grade A+")
elif marks >= 80:
    print("Grade A")
elif marks >= 70:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
elif marks >= 50:
    print("Grade D")
else:
    print("Fail")
```
