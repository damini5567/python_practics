#### nested if block

```python
num = 20
if num > 0:
    if num % 2 == 0:
        print("Positive Even number")
```

```python
age = 25
citizen = True
if age >= 18:
    if citizen:
        print("You can vote")
print("Thank you for checking whether you are eligible to vote.")
```
