## Basic Tuple Operations

### 1. Immutability

```python
tuple1=(10,20,30)
tuple1[0]=11
```

#### **** List    ******

```python
list1=[10,20,30]
list1[0]=11
print(list1)
```

### 2&3. Concatenation & Merging

```python
tuple1=(10,20,30)
tuple2=(40,50,60)
tuple3=tuple1+tuple2
print(tuple3)
```

### 4. Conversion

#### 1. string to tuple

```python
str1="Hello"
tuple1=tuple(str1)
print(tuple1)
```

#### 2. list to tuple

```python
list1=[10,20,30]
tuple1=tuple(list1)
print(tuple1)
```

#### 3. set to tuple

```python
set1={10,20,30,40,50}
tuple1=tuple(set1)
print(tuple1)
```

#### 4.dictionary to tuple

```python
dict1={
    "a":1,
    "b":2
}
tuple1=tuple(dict1)
print(tuple1)
```

#### 5.range() to tuple

```python
range1=range(1,11,1)
tuple1=tuple(range1)
print(tuple1)
```

#### 5.Aliasing

```python
tuple1=(10,20,30)
tuple2=tuple1
print(id(tuple1))
print(id(tuple2))
```

#### 6. Cloning

```python
tuple1=(10,20,30)
tuple2=tuple(tuple1)
#tuple2=tuple1+()
print(tuple1)
print(tuple2)
print(id(tuple1))
print(id(tuple2))
```

#### 7.Searching - in operator

```python
tuple1=(10,10.1,"Avantika",True,2+3j)
```

```python
print(10 in tuple1)
print("Avantika" in tuple1)
print(100 in tuple1)
```

#### 8. Identity - is operator

```python
tuple1=(10,20,30)
tuple2=tuple1
print(tuple1 is tuple2)
print(tuple2 is tuple1)
print(id(tuple1))
print(id(tuple2))
```

```python
tuple3=(40,50,60)
print(tuple1 is tuple3)
print(tuple3 is tuple1)
print(id(tuple1))
print(id(tuple3))
```

### 9.Comparision

```python
tuple1=(10,20,30,40)
tuple2=(10,20,50,60)
print(tuple1<tuple2)
print(tuple1>tuple2)
```

### 10. Emptiness

```python
tuple1=()
print(not tuple1)
```

```python
tuple2=(10,20,30)
print(not tuple2)
```
