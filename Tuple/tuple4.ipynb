## Tuple methods

### 1. index( )

```python
tuple1=(10,20,30,40,50)
print(tuple1.index(30))
print(tuple1.index(50))
# print(tuple1.index(100)) ->ValueError
```

### 2. count( )

```python
tuple1=(10,20,30,10,40,10)
print(tuple1.count(10))
print(tuple1.count(40))
```

### Neasted tuple

```python
tuple1=(10,20,30)
tuple2=(40,50,60)
tuple3=(tuple1,tuple2)
print(tuple3)
```
