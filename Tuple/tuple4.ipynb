## Tuple methods

### 1. index( ) - it returns occurance of the elements

```python
tuple1=(10,20,30,40,50)
print(tuple1.index(30))
print(tuple1.index(50))
# print(tuple1.index(100)) ->ValueError
```

### 2. count( ) - it returns how mant times single element occur in tuple

```python
tuple1=(10,20,30,10,40,10)
print(tuple1.count(10))
print(tuple1.count(40))
```

### Neasted tuple

```python
tuple1=(10,20,30)
tuple2=(40,50,60)
tuple3=(tuple1,tuple2)
print(tuple3)
```

```python
print(tuple3[0])

print(tuple3[1])
```

```python
print(tuple3[0][0])
print(tuple3[0][1])
print(tuple3[0][2])
```
