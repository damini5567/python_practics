## Built in functions in tuple

### 1. len( )

```python
tuple1=(10,20,30)
print(len(tuple1))
```

### 2. max( )

```python
tuple1=(10,20,30)
print(max(tuple1))
```

### 3.min( )

```python
tuple1=(10,20,30)
print(min(tuple1))
```

### 4.sum( )

```python
tuple1=(10,20,30)
print(sum(tuple1))
```

### 5.any( )

```python
tuple1=(0,0,0,1)
print(any(tuple1))
```

```python
tuple2=(0,0,0,0)
print(any(tuple2))
```

### 6. all( )

```python
tuple1=(10,20,30,40,"Avantika",10.1)
print(all(tuple1))
```

```python
tuple1=(10,20,30,40,"Avantika",0)
print(all(tuple1))
```
