## Tuple in Python

#### Define Tuple

```python
tuple1=(21,22,23,24,25)
print(tuple1)
print(type(tuple1))
```

### Properties Of Tuple

#### 1. Homogeneous tuple = store elements of same data type

```python
tuple1=(10,20,30,40,50)
print(tuple1)
print(type(tuple1))
```

#### 2. Heterogeneous tuple = store elements of different data type

```python
tuple2=("Avantika",1,0.1,3+2j,True)
print(tuple2)
```

#### Duplicates elements are allowed

```python
tuple3=(10,20,30,10,20)
print(tuple3)
print(type(tuple3))
```

#### Creating empty tuple

```python
tuple4=()
print(tuple4)
print(type(tuple4))
```

#### creating tuple with only one element

```python
tuple5=(10,)
print(tuple5)
print(type(tuple5))

tuple6=(50)
print(tuple6)
print(type(tuple6))

tuple7=(10,20)
print(tuple7)
print(type(tuple7))
```
