### Python Basic Data Types

#### 1) integer

```python
decimal_num=10
print(decimal_num)
print(type(decimal_num))
```

```python
binary_num=0b1010
print(binary_num)
```

```python
octal_num=10
print(octal_num)
```

```python
hexadecimal_num=0xA
print(hexadecimal_num)
```

#### 2) float

##### i) decimal notation

```python
pi=3.14
print(pi)
print(type(pi))
```

##### ii) exponential Notation

```python
var1=6.02e23
print(var1)
print(type(var1))
```

#### 3) complex

```python
var1=5+2j
print(var1)
print(var1.real)
print(var1.imag)
print(type(var1))
```

#### 4) boolean

```python
is_student = True
print(is_student)
```

#### 5) string

```python
city = "Pune"
print(city)
print(type(city))
```
