#### accepting two values and return its addition - Solution1

```python
num1=input("Enter first value:")  # 250
print(num1)
print(id(num1))
print(type(num1))

num1=int(num1)                    # "250" --> 250
print(num1)      
print(type(num1))

num2=input("Enter second value")  # 100
print(num2)
print(id(num2))
print(type(num2))

num2=int(num2)                    # "100" --> 100
print(num2)
print(type(num2))

result=num1+num2                  # 
print(type(result))
```
