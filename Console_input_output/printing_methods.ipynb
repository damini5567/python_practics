## FORMATTED OUTPUT

### 1) Normal Method

```python
name="Avantika"
age=25
print("Name:",name, "Age:",age)      
print("Name:",name,",","Age:",age)  
```

```python
print("My Name is:",name,",", "My Age is:",age)
```

```python
roll_no=32
address="pune"
print("My address is",address,"and roll_no is",roll_no)
# my address is pune and roll_no is 32
```

### 2) String Concatenation

```python
"hello"+"world"
```

```python
"hello"+str(10)
```

```python
"Welcome"+" "+str(10)+" "+"hello"+" "+"world"
```

```python
roll_no=32
address="pune"
roll_no=str(roll_no)        # 32 -> "32"
print("My address is "+address+" and roll_no is "+roll_no)
print("My address is", address, "and roll_no is", roll_no)
```

```python
"my address is" + " pune"+" and roll_no is "+"32"
```

```python
a="hello"
b="world"
c=a+" "+b
print(c)
```

### 3) format() Method

```python
roll_no=32
address="pune"
pin_code=412105

print("My address is: {} and roll_no is: {} and pin code is: {}".format(address,roll_no,pin_code))
# My address is: pune and roll_no is: 32 and pin code is: 412105
```

```python
name="Avantika"
age=25
print("Name:{}, Age:{}".format(name,age))  
```

### 4) f-string Method

```python
roll_no=32
address="pune"
pin_code=412105

print(f"My address is {address} and roll_no is {roll_no} and pin code is {pin_code}")
```
